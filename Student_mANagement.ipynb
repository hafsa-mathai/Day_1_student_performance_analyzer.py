```python
students={}
#Grade function
def grade(marks):
  if marks>=80:
    return 'A'
  elif marks>=70:
    return 'B'
  elif marks>=60:
    return 'C'
  elif marks>=50:
    return 'D'
  else:
    return 'F'

#add function
def add():
  roll=input("Enter Roll No:")
  name=input("Enter Name:")
  marks=int(input("Enter Marks:"))
  students[roll]=[name,marks,grade(marks)]
  print("Student added")

#update student
def update():
  roll=input("Enter Roll No:")
  if roll in students:
    name=input("Enter Name:")
    marks=int(input("Enter Marks:"))
    students[roll]=[name,marks,grade(marks)]
    print("updated")
  else:
    print("Student not found")

#delete student
def delete():
  roll=input("Enter Roll No:")
  if roll in students:
    del students[roll]
    print("deleted")
  else:
    print("Student not found")

#search student
def search():
  roll=input("Enter Roll No:")
  if roll in students:
    print("Record",students[roll])
  else:
    print("Student not found")


#delete student
def delete():
  roll=input("Enter Roll No:")
  if roll in students:
    del students[roll]
    print("deleted")
  else:
    print("Student not found")

 # Menu
while True:
    print(students)
    print("\n1 Add  2 Update  3 Delete  4 Search  5 Exit")
    choice = input("Enter choice: ")

    if choice == "1":
        add()
    elif choice == "2":
        update()
    elif choice == "3":
        delete()
    elif choice == "4":
        search()
    elif choice == "5":
        print("Program End")
        break
    else:
        print("Invalid choice")






```

we make that little student management program that include the function of add , delete , search ,update , grade
