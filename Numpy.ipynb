#Array vs list

|Array||list|
|---|-|---|
 |Same data type|-|Different data type|
 |contigous memory location|-|Not contiguous memory location|
 |static and can't be resized|-|mutable resized and modified easily|
 |Support element wise operation|-|do not support element wise operation|
 |takes less space in memory|-|more space in memory|

```python
import numpy as np
x=[[1,2,3],[4,5,6],[7,1,0]]
ar=np.array(x)
print(ar.sum(axis=0))
print(ar.sum(axis=1))
print(ar.T)
print(ar.flat)
for item in ar.flat:
  print(item)

print(ar.ndim)
print(ar.size)
```

```python
import numpy as np
one=np.array([1,3,56,78,65])
one.argmax() #tells wher maximum value exists
print(one.argmin())
print(one.argsort()) #gives indices on which the array is sorted
```

```python
import numpy as np
x=[[1,2,3],[4,5,6],[7,1,0]]
ar=np.array(x)
ar1=np.array([[1,2,1],
              [4,0,6],
              [8,1,0]])
print("Sum:",ar+ar1)
print("Product:",ar*ar1)
print("Difference:",ar-ar1)
print("Division:",ar/ar1)
print(np.sqrt(ar))
np.where(ar>5)
np.count_nonzero(ar)
```
