```python
#Banking system
#create account
account={}
print("Welcome to my mini bank.")
def create_account():
  name=input("Whats your name bro::")
  pin=input("Lets create some strong pin:")
  balance=float(input("Let me see are you rich or broke , lets tell me your balance::"))
  account['name']=name
  account['pin']=pin
  account['balance']=balance
  account['history']=[]  #just created an emty list to store transaction history later
  account['history'].append("Account created succes")


  print("Account Created")
  print(f"Name:{name}")
  print(f"Pin:{pin}")
  print(f"Balance:{balance}")

  return account



#Deposit
def deposit(account):
  amount=float(input("Enter amount to deposit:"))
  account['balance']=account['balance']+amount
  account['history'].append('Deposited:'+str(amount))
  print("After deposit")
  print("Balance",account['balance'])

def withdraw(account):
    print("\n--- STEP 3: WITHDRAW ---")
    amount = float(input("Enter amount to withdraw: "))

    if amount > account["balance"]:
        print("Error: Insufficient balance!")
    else:
        account["balance"] -= amount  # Typo fixed here
        account["history"].append(f"Withdrawn: {amount}")

    print("\n--- AFTER WITHDRAW ---")
    print("Balance:", account["balance"])

def show_history(account):
    print("\n--- STEP 4: TRANSACTION HISTORY ---")
    for record in account["history"]:
        print(f"- {record}")

def main():
    # Calling the functions step by step
    my_account = create_account()

    deposit(my_account)

    withdraw(my_account)

    show_history(my_account)

# Run the program
if __name__ == "__main__":
    main()

```
